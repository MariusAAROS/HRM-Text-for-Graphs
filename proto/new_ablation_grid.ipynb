```python
h = (1, 3, 6, 1, 3, 5, 6, 2, 3, 4, 6, 1, 2, 4, 1, 2) + (3, 4, 2, 4, 1) # new + reused
l = (3, 3, 3, 6, 6, 6, 6, 12, 12, 12, 12, 1, 1, 1, 24, 24) + (3, 3, 6, 6, 12) # new + reused
```

```python
print(sorted([[l_/h_, l_, h_] for l_, h_ in zip(l, h)], key=lambda x: x[0]))
```

```python
print([h_*l_ for h_, l_ in zip(h, l)])
```

```python
# position L and H on a 2D grid 
import matplotlib.pyplot as plt

# color runs based on both their L/H ratio and their product
colors = [l_/h_ * l_*h_ for l_, h_ in zip(l, h)]

# print L/H and L*H for each run on the side of each point
for l_, h_ in zip(l, h):
    plt.text(h_, l_, f"{l_/h_:.2f}, {l_*h_:.2f}", fontsize=8)

plt.scatter(h, l, c=colors, cmap='Reds')
plt.colorbar(label='L/H')
plt.xlabel('H')
plt.ylabel('L')
plt.show()
```
